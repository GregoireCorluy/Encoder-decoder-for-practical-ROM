## Explore CFLF dataset of Malik

Conclusions:
- Temperature has 400 observations less and probably has one simulation missing since the jump in values are happening at the same index

```python
path = "data-files/"
name_mf = "Malik-mf-CFLF-2022.csv"
name_T = "Malik-T-CFLF-2022.csv"
name_state_space = "Malik-state-space-CFLF-2022.csv"

data_mf = pd.read_csv(path + name_mf)
data_T = pd.read_csv(path + name_T)
data_state_space = pd.read_csv(path + name_state_space)
```

```python
data_mf
```

```python
plt.scatter(data_mf, data_state_space["CO2"], s = 1)
plt.show()
```

```python
species = "H"

plt.figure(figsize=(12, 4))  # width=12 inches, height=4 inches
plt.plot(data_T[0:450]/data_T[0:450].max())
plt.plot(data_state_space[species][0:450]/data_state_space[species][0:450].max())
plt.show()

idx_T = data_T.iloc[:450].idxmax()
idx_species = data_state_space[species].iloc[:450].idxmax()

print(idx_T, idx_species)
```

```python
species = "CH4"

plt.figure(figsize=(12, 4))  # width=12 inches, height=4 inches
plt.plot(data_T[0:450]/data_T[0:450].max())
plt.plot(data_state_space[species][0:450]/data_state_space[species][0:450].max())
plt.show()

print(data_T[399:402])
print(data_state_space[species][399], data_state_space[species][400], data_state_space[species][401])
```
